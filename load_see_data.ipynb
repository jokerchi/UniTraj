```python
import pandas as pd
import numpy as np
import folium
import random
import os

os.makedirs("visualization", exist_ok=True)
pd.set_option("display.max_colwidth", None)   # 不截断单元格内容
pd.set_option("display.max_columns", None)    # 显示所有列
pd.set_option("display.width", 200)           # 调整行宽
```

```python
worldtrace = pd.read_pickle('./data/worldtrace_sample.pkl')
```

```python
worldtrace.head(2)
```

```python
worldtrace.size
```

```python
# random select 16 trajectories from worldtrace and draw it
worldtrace_sample = worldtrace.sample(100)
trajectory = worldtrace_sample['trajectory']
trajectory = trajectory.apply(lambda x: np.array(x))
trajectory = trajectory.values

m = folium.Map(location=trajectory[0][0], zoom_start=13)
for i in range(len(trajectory)):
    folium.Marker(list(trajectory[i][0])).add_to(m)
    color = '#' + ''.join([random.choice('0123456789ABCDEF') for j in range(6)])
    folium.PolyLine(trajectory[i], color = color,weight=10).add_to(m)

m.save('./visualization/sample.html')
m
```
